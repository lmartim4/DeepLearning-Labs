### Section 1. Introduction
#### *What is Time-series Forecasting?*
In machine learning, **time-series forecasting** refers to predicting future values of a variable based on its historical observations. Rather than treating each data point independently, time-series models capture **temporal dependencies**, i.e., how past values influence future outcomes. This requires learning patterns such as trends, seasonal cycles, and irregular fluctuations from sequential data.


#### *Why is Time-series Forecasting interesting?*
- ✅ **Foundation of practical applications**: Stock price prediction, demand forecasting, weather prediction, energy consumption, anomaly detection, all of them rely on accurate time-series models.

- 🧠 **Learning complex temporal patterns**: Deep learning excels where traditional methods (e.g., ARIMA) fail, capturing non-linear dependencies and long-range temporal correlations through LSTMs, Transformers, and attention mechanisms.

- 🔄 **Bridging past and future**: A strong time-series model helps systems understand "what happened before" and "what comes next", which is not only crucial for decision-making in finance, healthcare, and autonomous systems, but also **unveiling cosmic secrets**! In astrophysics, research from detecting gravitational wave signals buried in LIGO detector noise, to predicting stellar light curves and identifying transient events like supernovae or gamma-ray bursts, time-series deep learning helps us listen to the universe and understand its most violent and mysterious moments.

- 🚀 **Real-world impact**: Even 1-2% improvements in forecasting accuracy translate to millions in cost savings for retailers and energy companies, or life-saving interventions in healthcare.

### Section 4: Classical Methods - Distance-Based Time Series Comparison

#### *What is Distance-Based Time Series Analysis?*

**Distance-based approaches** are classical machine learning methods that compare time series by measuring similarity using distance metrics. Rather than learning patterns from data, these methods rely on pre-defined distance functions to compare sequences directly. For time series that may be shifted or scaled versions of each other, specialized metrics like **Dynamic Time Warping (DTW)** are used instead of simple Euclidean distance.

The core idea is elegantly simple but effective:

- **Intuitive distance metrics**: Euclidean distance works for aligned sequences, but DTW allows flexible alignment by "warping" the time axis to find optimal matching.

- **No training required**: These methods are parameter-free and don't require optimization. 

- **Interpretability**: The similar time series are explicit, so that we can directly see which training example the test series matched with.

- **Baseline strength**: Despite their simplicity, distance-based methods could also outperform sophisticated deep learning on small, well-behaved datasets.

```python
# Use the already prepared data from Section 3
# (flux_norm and time_clean are already available)


# ===================================================================
# SUPER SIMPLE: Create trivially different classes
# ===================================================================




print("\nCreating TRIVIALLY separable dataset...\n")

n_samples = len(transit_windows)
n_per_class = n_samples // 3

# Class 0: Baseline + small noise
class0 = np.ones((n_per_class, 100)) + np.random.normal(0, 0.001, (n_per_class, 100))

# Class 1: Baseline + 0.01 offset + small noise (clearly higher)
class1 = 1.01 * np.ones((n_per_class, 100)) + np.random.normal(0, 0.001, (n_per_class, 100))

# Class 2: Baseline - 0.01 offset + small noise (clearly lower)
class2 = 0.99 * np.ones((n_per_class, 100)) + np.random.normal(0, 0.001, (n_per_class, 100))

transit_data_section = np.vstack([class0, class1, class2])
transit_labels_section = np.hstack([
    np.zeros(n_per_class, dtype=int),
    np.ones(n_per_class, dtype=int),
    2 * np.ones(n_per_class, dtype=int)
])

print(f"✓ Trivial dataset created:")
print(f"  Class 0 mean: {class0.mean():.6f}")
print(f"  Class 1 mean: {class1.mean():.6f}")
print(f"  Class 2 mean: {class2.mean():.6f}\n")

# This should be 100% separable - if LSTM still fails, there's something else wrong



# ===================================================================
# STEP 2: PREPARE TRAINING/TEST SPLIT
# ===================================================================

X = transit_data_section
y = transit_labels_section

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=RANDOM_SEED, stratify=y
)

print(f"Train/Test split:")
print(f"  Train set: {X_train.shape}")
print(f"  Test set: {X_test.shape}")
print(f"  Train distribution: {np.bincount(y_train)}")
print(f"  Test distribution: {np.bincount(y_test)}\n")

```

```python
# ===================================================================
# STEP 3: KNN WITH EUCLIDEAN DISTANCE
# ===================================================================

print("="*70)
print("METHOD 1: KNN with Euclidean Distance")
print("="*70 + "\n")

knn_euclidean = KNeighborsClassifier(n_neighbors=5, metric='euclidean')
knn_euclidean.fit(X_train, y_train)

y_pred_euclidean = knn_euclidean.predict(X_test)
score_euclidean = knn_euclidean.score(X_test, y_test)

print(f"Test Accuracy: {score_euclidean:.4f}\n")

# Detailed per-class metrics
cm_euclidean = confusion_matrix(y_test, y_pred_euclidean)
print("Confusion Matrix (Euclidean):")
print(cm_euclidean)
print()

# ===================================================================
# STEP 4: KNN WITH DYNAMIC TIME WARPING (DTW)
# ===================================================================

print("="*70)
print("METHOD 2: KNN with Dynamic Time Warping (DTW)")
print("="*70 + "\n")

def dtw_distance(x, y):
    """
    Compute Dynamic Time Warping distance between two sequences.
    
    DTW allows flexible alignment by "warping" the time axis to find
    the optimal matching between sequences.
    """
    n, m = len(x), len(y)
    
    # Initialize DTW cost matrix
    dtw_matrix = np.full((n + 1, m + 1), np.inf)
    dtw_matrix[0, 0] = 0
    
    # Fill cost matrix using dynamic programming
    for i in range(1, n + 1):
        for j in range(1, m + 1):
            # Cost is absolute difference between points
            cost = abs(x[i-1] - y[j-1])
            
            # Recurrence relation: choose path with minimum cost
            dtw_matrix[i, j] = cost + min(
                dtw_matrix[i-1, j],      # insertion
                dtw_matrix[i, j-1],      # deletion
                dtw_matrix[i-1, j-1]     # match
            )
    
    return dtw_matrix[n, m]

print("Computing DTW distances (this may take few minutes)...\n")


# Precompute DTW distance matrix between all test and train samples
n_test = len(X_test)
n_train = len(X_train)
dtw_distances = np.zeros((n_test, n_train))

for i in range(n_test):
    for j in range(n_train):
        dtw_distances[i, j] = dtw_distance(X_test[i], X_train[j])
    
    if (i + 1) % 10 == 0:
        print(f"  Processed {i+1}/{n_test} test samples")

print(f"\n✓ DTW distance matrix computed: {dtw_distances.shape}\n")

# Classify using nearest neighbor in DTW space
y_pred_dtw = np.zeros(n_test, dtype=int)
for i in range(n_test):
    nearest_idx = np.argmin(dtw_distances[i])
    y_pred_dtw[i] = y_train[nearest_idx]

score_dtw = np.mean(y_pred_dtw == y_test)

print(f"Test Accuracy: {score_dtw:.4f}\n")

# Detailed per-class metrics
cm_dtw = confusion_matrix(y_test, y_pred_dtw)
print("Confusion Matrix (DTW):")
print(cm_dtw)
print()
```

<span style="color:orange">**Questions:**</span> 
 - Q1: Why does DTW outperform Euclidean distance?
 - Q2: What do the spikes in the transit plot represent?
 - Q3: Why is DTW much slower than Euclidean distance?


<span style="color:green">**Answers:**</span>
 - **A1:** Euclidean distance is rigid, it compares point-by-point and penalizes small time shifts heavily. DTW flexibly aligns sequences by warping the time axis, allowing it to match transits that occur at slightly different times. For astronomical data with phase jitter and noise, this flexibility is crucial.
- **A2:** The central dip is the actual exoplanet transit. The surrounding spikes are **noise** from stellar activity, photometric uncertainty, or instrument artifacts. 
- **A3:** Euclidean distance is O(n) per pair. DTW requires dynamic programming, i.e., O(n²) per pair, to find the  optimal temporal alignment. At scale, this becomes prohibitive, motivating the deep learning approaches in following sections, which learn efficient representations without explicit alignment.

**For instructors only: Understanding the middle plot (Example Transit Event)**

**What is being shown:**
- **Blue line with dots**: The actual light curve of a single exoplanet transit event extracted from the test set. Each dot is one flux measurement (brightness of the star).

- **Red dashed line**: The baseline (flux = 1.0, meaning full stellar brightness with no obstruction).

- **Green shaded region**: The expected transit window, where we anticipate the exoplanet to cross in front of its star.

**Reading the plot:**
1. The y-values mostly hover around 1.0 (baseline brightness).
2. In the middle (samples ~45-55), there's a small **dip to ~0.998**, and this is the exoplanet transit (the planet blocks ~0.2% of starlight).
3. Around the transit, there are **spiky noise fluctuations**, these are instrumental/atmospheric noise.
4. The rest of the signal is mostly noise with no clear transit features.

### Section 5: Deep Learning - LSTM for Time Series Classification

#### What is an LSTM for Time Series?

**Long Short-Term Memory (LSTM)** networks are recurrent neural networks designed to learn temporal dependencies in sequential data. Unlike distance-based methods that compare entire sequences at once, LSTMs process sequences step-by-step, maintaining an internal "memory" (hidden state) that captures relevant information from past observations.

<span style="color:blue">**Tasks:**</span> Fill in the indicated blanks to build and train an LSTM classifier:

- Prepare the data into sequences suitable for LSTM input (reshape and batch data);
- Define an LSTM-based neural network architecture;
- Train the model on the training set and evaluate on the test set.


```python
# ===================================================================
# STEP 1: PREPARE DATA FOR LSTM
# ===================================================================
# 👉 [STUDENTS CAN IMPLEMENT THIS PART]
# Hint: Convert numpy arrays to PyTorch tensors and reshape for LSTM

# From previous section: X_train, X_test, y_train, y_test already exist
# LSTM expects shape: (batch_size, sequence_length, num_features)
# Currently X_train shape is (n_samples, 100) → need (n_samples, 100, 1)

X = transit_data_section
y = transit_labels_section

# Train/test split (same as Section 4)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=RANDOM_SEED, stratify=y
)

# Convert to PyTorch tensors: (batch_size, sequence_length, num_features)
X_train_lstm = torch.FloatTensor(X_train).unsqueeze(-1)  # (N_train, 100, 1)
X_test_lstm = torch.FloatTensor(X_test).unsqueeze(-1)    # (N_test, 100, 1)
y_train_lstm = torch.LongTensor(y_train.astype(int))
y_test_lstm = torch.LongTensor(y_test.astype(int))

print(f"Data prepared for LSTM:")
print(f"  Train shape (LSTM format): {X_train_lstm.shape}  [batch, sequence, features]")
print(f"  Test shape:  {X_test_lstm.shape}")
print(f"  Train labels: {y_train_lstm.shape}")
print(f"  Num classes: 3 (Pristine, Moderate, Degraded)\n")

# Create data loaders
batch_size = 4  # Small batch size for small dataset
train_dataset = TensorDataset(X_train_lstm, y_train_lstm)
train_loader = DataLoader(train_dataset, batch_size=batch_size, shuffle=True)

print(f"Data Loader created:")
print(f"  Batch size: {batch_size}")
print(f"  Total batches per epoch: {len(train_loader)}\n")

# print(y_train_lstm)


```

```python
# ===================================================================
# STEP 2: DEFINE LSTM MODEL
# ===================================================================
# 👉 [STUDENTS CAN IMPLEMENT THIS PART]
# Hint: Design an LSTM architecture with:
#       - Input dimension: 1 (single flux value per timestep)
#       - LSTM hidden dimension: suggest 64 or 128
#       - Output: 3 classes

class LSTMClassifier(nn.Module):
    def __init__(self, input_size=1, hidden_size=256, num_layers=3, 
                 num_classes=3, dropout=0.5):
        super(LSTMClassifier, self).__init__()
        
        # Bidirectional LSTM (processes sequence forward AND backward)
        self.lstm = nn.LSTM(
            input_size=input_size,
            hidden_size=hidden_size,
            num_layers=num_layers,
            batch_first=True,
            dropout=dropout if num_layers > 1 else 0,
            bidirectional=True  # ← KEY: learn from both directions
        )
        
        # After bidirectional LSTM, output is 2*hidden_size
        lstm_output_size = hidden_size * 2
        
        # Heavy classification head with multiple layers
        self.fc1 = nn.Linear(lstm_output_size, 256)
        self.bn1 = nn.BatchNorm1d(256)
        self.relu1 = nn.ReLU()
        self.dropout1 = nn.Dropout(dropout)
        
        self.fc2 = nn.Linear(256, 128)
        self.bn2 = nn.BatchNorm1d(128)
        self.relu2 = nn.ReLU()
        self.dropout2 = nn.Dropout(dropout)
        
        self.fc3 = nn.Linear(128, num_classes)     

        
    def forward(self, x):
        # x: (batch, seq_len, input_size)
        lstm_out, (h_n, c_n) = self.lstm(x)
        # lstm_out: (batch, seq_len, 2*hidden_size) from bidirectional
        
        # Use last timestep output
        last_output = lstm_out[:, -1, :]  # (batch, 2*hidden_size)
        
        # Classification head
        x = self.fc1(last_output)
        x = self.bn1(x)
        x = self.relu1(x)
        x = self.dropout1(x)
        
        x = self.fc2(x)
        x = self.bn2(x)
        x = self.relu2(x)
        x = self.dropout2(x)
        
        logits = self.fc3(x)
        return logits
```

```python
# ===================================================================
# STEP 3: DEFINE LOSS AND OPTIMIZER, TRAIN THE MODEL
# ===================================================================

### TO BE FILLED: Initialize model
model = LSTMClassifier(
    input_size=1,
    hidden_size=256,  
    num_layers=3,   
    num_classes=3,
    dropout=0.5
)
model = model.to(device)


### TO BE FILLED:
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr=0.01) 

num_epochs = 200
best_val_acc = 0
early_stop_counter = 0
early_stop_patience = 30

train_losses = []
val_accs = []

print("Training LSTM...\n")

for epoch in range(num_epochs):
    model.train()
    train_loss = 0
    
    for batch_X, batch_y in train_loader:
        batch_X = batch_X.to(device)
        batch_y = batch_y.to(device)
        
        outputs = model(batch_X)
        loss = criterion(outputs, batch_y)
        
        optimizer.zero_grad()
        loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
        optimizer.step()
        
        train_loss += loss.item()
    
    avg_train_loss = train_loss / len(train_loader)
    train_losses.append(avg_train_loss)
    
    # Validation
    model.eval()
    with torch.no_grad():
        val_outputs = model(X_test_lstm.to(device))
        val_preds = torch.argmax(val_outputs, dim=1).cpu().numpy()
        val_acc = np.mean(val_preds == y_test.astype(int))
        val_accs.append(val_acc)

    # Early stopping
    if val_acc > best_val_acc:
        best_val_acc = val_acc
        early_stop_counter = 0
        best_model_state = model.state_dict()
    else:
        early_stop_counter += 1
    
    if (epoch + 1) % 10 == 0:
        print(f"Epoch [{epoch+1:3d}] Loss: {avg_train_loss:.4f}, Val Acc: {val_acc:.4f}")
    
    # if early_stop_counter >= early_stop_patience:
    #     print(f"Early stopping at epoch {epoch+1}")
    #     break

print(f"\n✓ Best validation accuracy: {best_val_acc:.4f}\n")

print("\n✓ Training complete!\n")
```

```python
# ===================================================================
# STEP 5: COMPARE RESULTS
# ===================================================================

model.eval()

with torch.no_grad():
    test_logits = model(X_test_lstm.to(device))
    test_preds = torch.argmax(test_logits, dim=1).cpu().numpy()
    test_probs = torch.nn.functional.softmax(test_logits, dim=1).cpu().numpy()

test_acc = np.mean(test_preds == y_test.astype(int))
cm_lstm = confusion_matrix(y_test.astype(int), test_preds)

print("="*70)
print("LSTM TEST RESULTS")
print("="*70)
print(f"Test Accuracy: {test_acc:.4f}\n")
```

### Section 6: Transformers with Self-Attention

#### What is a Transformer for Time Series?

**Transformer networks** use **self-attention mechanisms** to process sequences, allowing each 
timestep to directly attend to all other timesteps. Unlike RNNs (LSTM/GRU) that process sequences 
sequentially, Transformers compute all time relationships in parallel, making them both more 
efficient and potentially more powerful at capturing long-range temporal dependencies.

The key insight: Instead of asking "what happened at the previous timestep?", Transformers ask 
"which parts of the entire sequence are most relevant for understanding this moment?"


#### How Transformers Learn Transit Patterns

On our exoplanet transit data with temporal shifts, Transformers can:
- **Attend to the dip location**: Learn that all timesteps around the transit dip are highly relevant
- **Distinguish shift patterns**: Use attention to identify whether the dip is centered, left-shifted, 
  or right-shifted
- **Handle noise robustly**: Downweight noisy outliers while focusing on consistent transit features

#### Tasks

Fill in the indicated blanks to implement a Transformer classifier:

- Positional encoding: Add temporal position information (since Transformers don't inherently 
  understand position)
- Multi-head self-attention: Build an attention layer with multiple representation heads
- Feed-forward network: Add non-linearity to the attention output
- Classification head: Design the output layer for 3-class prediction
- Training & evaluation: Compare with distance-based and LSTM baselines


```python
# ===================================================================
# STEP 1: PREPARE DATA FOR TRANSFORMER
# ===================================================================

X = transit_data_section
y = transit_labels_section

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=RANDOM_SEED, stratify=y
)

# Convert to PyTorch tensors: (batch_size, sequence_length, num_features)
X_train_tf = torch.FloatTensor(X_train).unsqueeze(-1)  # (N_train, 100, 1)
X_test_tf = torch.FloatTensor(X_test).unsqueeze(-1)    # (N_test, 100, 1)
y_train_tf = torch.LongTensor(y_train.astype(int))
y_test_tf = torch.LongTensor(y_test.astype(int))

# ===================================================================
# STEP 2: CREATE DATA LOADERS
# ===================================================================

batch_size = 8
train_dataset = TensorDataset(X_train_tf, y_train_tf)
train_loader = DataLoader(train_dataset, batch_size=batch_size, shuffle=True)

print(f"Data Loader created:")
print(f"  Batch size: {batch_size}")
print(f"  Total batches per epoch: {len(train_loader)}\n")
```

```python
# ===================================================================
# STEP 4: DEFINE TRANSFORMER MODEL
# ===================================================================
# 👉 [STUDENTS CAN IMPLEMENT THIS PART]
# Hint: Build positional encoding, attention layers, and classification head

class PositionalEncoding(nn.Module):
    """Add positional information to sequence tokens."""
    
    def __init__(self, d_model, max_len=100):
        super(PositionalEncoding, self).__init__()
        
        # Create positional encoding matrix
        pe = torch.zeros(max_len, d_model)
        position = torch.arange(0, max_len, dtype=torch.float).unsqueeze(1)
        
        # Compute frequencies using sine/cosine basis functions
        div_term = torch.exp(torch.arange(0, d_model, 2).float() * 
                            (-math.log(10000.0) / d_model))
        
        pe[:, 0::2] = torch.sin(position * div_term)
        if d_model % 2 == 1:
            pe[:, 1::2] = torch.cos(position * div_term[:-1])
        else:
            pe[:, 1::2] = torch.cos(position * div_term)
        pe = pe.unsqueeze(0)  # (1, max_len, d_model)
        self.register_buffer('pe', pe)
    
    def forward(self, x):
        """
        Args:
            x: (batch_size, seq_len, d_model)
        Returns:
            x + positional encoding
        """
        return x + self.pe[:, :x.size(1), :]

class TransformerBlock(nn.Module):
    """Single Transformer block: Multi-head attention + Feed-forward."""
    
    def __init__(self, d_model, num_heads, d_ff, dropout=0.1):
        super(TransformerBlock, self).__init__()
        
        # Multi-head self-attention
        self.attention = nn.MultiheadAttention(
            embed_dim=d_model,
            num_heads=num_heads,
            dropout=dropout,
            batch_first=True
        )
        
        # Feed-forward network
        self.ff = nn.Sequential(
            nn.Linear(d_model, d_ff),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(d_ff, d_model)
        )
        
        # Layer normalization
        self.norm1 = nn.LayerNorm(d_model)
        self.norm2 = nn.LayerNorm(d_model)
        
        # Dropout
        self.dropout = nn.Dropout(dropout)
        
    def forward(self, x):
        """
        Args:
            x: (batch_size, seq_len, d_model)
        Returns:
            output: (batch_size, seq_len, d_model)
        """
        # Self-attention with residual connection and layer norm
        attn_out, attn_weights = self.attention(x, x, x)
        x = self.norm1(x + self.dropout(attn_out))
        
        # Feed-forward with residual connection and layer norm
        ff_out = self.ff(x)
        x = self.norm2(x + self.dropout(ff_out))
        
        return x

class TransitTransformer(nn.Module):
    """Transformer for exoplanet transit time series classification."""
    
    def __init__(self, input_size=1, d_model=64, num_heads=4, 
                 num_layers=2, d_ff=256, num_classes=3, dropout=0.1):
        super(TransitTransformer, self).__init__()
        
        # Input projection: expand flux to embedding dimension
        self.input_proj = nn.Linear(input_size, d_model)
        
        # Positional encoding
        self.pos_encoding = PositionalEncoding(d_model, max_len=100)
        
        # Stack of Transformer blocks
        self.transformer_blocks = nn.ModuleList([
            TransformerBlock(d_model, num_heads, d_ff, dropout)
            for _ in range(num_layers)
        ])
        
        # Classification head
        self.fc1 = nn.Linear(d_model, 128)
        self.relu = nn.ReLU()
        self.dropout = nn.Dropout(dropout)
        self.fc2 = nn.Linear(128, num_classes)
        
    def forward(self, x):
        """
        Args:
            x: (batch_size, seq_len, input_size)
        Returns:
            logits: (batch_size, num_classes)
        """
        # Project input to embedding dimension
        x = self.input_proj(x)  # (batch, seq_len, d_model)
        
        # Add positional encoding
        x = self.pos_encoding(x)  # (batch, seq_len, d_model)
        
        # Pass through Transformer blocks
        for block in self.transformer_blocks:
            x = block(x)  # (batch, seq_len, d_model)
        
        # Global average pooling: aggregate over time
        x = torch.mean(x, dim=1)  # (batch, d_model)
        
        # Classification layers
        x = self.fc1(x)
        x = self.relu(x)
        x = self.dropout(x)
        logits = self.fc2(x)  # (batch, num_classes)
        
        return logits
```

```python
# Initialize model
model = TransitTransformer(
    input_size=1,
    d_model=64,
    num_heads=4,
    num_layers=2,
    d_ff=256,
    num_classes=3,
    dropout=0.1
)
model = model.to(device)

criterion = nn.CrossEntropyLoss()
optimizer = optim.AdamW(model.parameters(), lr=0.001, weight_decay=1e-4)

num_epochs = 150
early_stop_patience = 25
early_stop_counter = 0
best_val_acc = 0

train_losses = []
val_accs = []

print(f"Training for up to {num_epochs} epochs...\n")

for epoch in range(num_epochs):
    # ==================== TRAIN ====================
    model.train()
    train_loss = 0
    train_correct = 0
    train_total = 0
    
    for batch_X, batch_y in train_loader:
        batch_X = batch_X.to(device)
        batch_y = batch_y.to(device)
        
        # Forward pass
        outputs = model(batch_X)
        loss = criterion(outputs, batch_y)
        
        # Backward pass
        optimizer.zero_grad()
        loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)
        optimizer.step()
        
        train_loss += loss.item()
        _, predicted = torch.max(outputs.data, 1)
        train_correct += (predicted == batch_y).sum().item()
        train_total += batch_y.size(0)
    
    avg_train_loss = train_loss / len(train_loader)

    train_acc = train_correct / train_total
    train_losses.append(avg_train_loss)
    
    # ==================== VALIDATION ====================
    model.eval()
    with torch.no_grad():
        val_outputs = model(X_test_tf.to(device))
        val_preds = torch.argmax(val_outputs, dim=1).cpu().numpy()
        val_acc = np.mean(val_preds == y_test.astype(int))
        val_accs.append(val_acc)
    
    # Early stopping (no scheduler.step() call)
    if val_acc > best_val_acc:
        best_val_acc = val_acc
        early_stop_counter = 0
        best_model_state = model.state_dict()
    else:
        early_stop_counter += 1
    
    if (epoch + 1) % 30 == 0:
        print(f"Epoch [{epoch+1:3d}/{num_epochs}] Train Loss: {avg_train_loss:.4f}, "
              f"Train Acc: {train_acc:.4f}, Val Acc: {val_acc:.4f}")
    
    # if early_stop_counter >= early_stop_patience:
    #     print(f"\nEarly stopping triggered at epoch {epoch+1}")
    #     break
```
